##1. Creating External location to access files azure storage

```sql
create external location if not exists dea_course_ext_dl_circuitbox
url 'abfss://circuitbox@de2026storag.dfs.core.windows.net/'
with (storage credential demo_storage)
COMMENT 'External Location for CircuitBox data lakehouse'
```

```sql
%fs ls 'abfss://circuitbox@de2026storag.dfs.core.windows.net/'
```

##2. Create Catalog for this project

```sql
create catalog if not exists circuitbox
    managed location 'abfss://circuitbox@de2026storag.dfs.core.windows.net/'
    comment 'CircuitBox data lakehouse Catalog';
```

```sql
use catalog circuitbox;

create schema if not exists landing
    managed location 'abfss://circuitbox@de2026storag.dfs.core.windows.net/landing';

create schema if not exists lakehouse
    managed location 'abfss://circuitbox@de2026storag.dfs.core.windows.net/lakehouse';
```

##3.Create Schemas for this project

```sql
show schemas
```

##4. Create volume for this project

```sql
use catalog circuitbox;
use schema landing;

create external volume if not exists operational_data
    location 'abfss://circuitbox@de2026storag.dfs.core.windows.net/landing/operational_data'
    comment 'Operational Data Volume';
```

```sql
%fs ls /Volumes/circuitbox/landing/operational_data
```
